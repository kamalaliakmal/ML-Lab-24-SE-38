## Task 3 · Randomness and reproducibility
- **(a)** Set `np.random.seed(2026)`, then generate one random number between 0 and 1.
- **(b)** Generate 25 numbers from the standard normal distribution.
- **(c)** Set `np.random.seed(42)` and roll 10,000 Ludo dice with `np.random.randint`. Count each face with `np.unique(..., return_counts=True)`.
- **(d)** Estimate P(six) and compare it with 1/6.
- **(e)** Count how many times two sixes came in a row (no loops).

```python
# T3: Randomness (a–e)
# Seed 2026: one uniform random value and 25 standard-normal values
np.random.seed(2026)
one_rand = np.random.rand()
twenty_five_randn = np.random.randn(25)
print("T3a one rand value:", one_rand)
print("T3b 25 randn values:", twenty_five_randn)

# Seed 42: simulate 10,000 Ludo dice rolls
np.random.seed(42)
rolls = np.random.randint(1, 7, size=10_000)
faces, face_counts = np.unique(rolls, return_counts=True)
probability_six = np.mean(rolls == 6)

# Consecutive double sixes: count adjacent pairs where both rolls are 6
double_sixes = np.sum((rolls[:-1] == 6) & (rolls[1:] == 6))

print("T3c face counts:", dict(zip(faces, face_counts)))
print("T3d estimated P(six):", probability_six)
print("T3e consecutive double-six pairs:", double_sixes)
```

## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:

```python
# T7: Climate (a–h)
print("T7a tmax shape:", tmax.shape)

yearly_mean_per_city = tmax.mean(axis=1)
hottest_month_index = tmax.argmax(axis=1)
hottest_month_per_city = months[hottest_month_index]
five_city_monthly_average = tmax.mean(axis=0)

# A month is listed if at least one city exceeds 35°C in that month.
months_any_city_above_35 = months[(tmax > 35).any(axis=0)]

lahore_fahrenheit = tmax[0] * 9 / 5 + 32

# July, August, September are columns 6, 7, 8 (stop 9 is excluded).
monsoon_rain = rain[:, 6:9].sum(axis=1)
wettest_monsoon_city = cities[monsoon_rain.argmax()]
monsoon_share_percent = monsoon_rain / rain.sum(axis=1) * 100

print("T7b yearly mean per city (°C):", np.round(yearly_mean_per_city, 2))
print("T7c hottest month per city:", dict(zip(cities, hottest_month_per_city)))
print("T7d five-city monthly average (°C):", np.round(five_city_monthly_average, 2))
print("T7e months where at least one city is above 35°C:", months_any_city_above_35)
print("T7f Lahore temperatures (°F):", np.round(lahore_fahrenheit, 2))
print("T7g monsoon rainfall Jul–Sep (mm):", dict(zip(cities, np.round(monsoon_rain, 1))))
print("T7h wettest monsoon city:", wettest_monsoon_city)
print("T7h monsoon share of annual rain (%):", dict(zip(cities, np.round(monsoon_share_percent, 2))))
```

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9: Grace marks (a–d)
# REQUIRED: `marks` must be the exact 10-value array from the starter notebook.
# Do not replace it with made-up marks.
if "marks" not in globals():
    raise NameError("Paste/run the original marks data cell from Lab02_Starter.ipynb before Task 9.")

marks = np.asarray(marks)
marks_after_grace = np.clip(marks + 5, 0, 100)

# Assumption: passing mark is 50/100. Change this if your course uses another rule.
PASS_MARK = 50
pass_percent = np.mean(marks_after_grace >= PASS_MARK) * 100
passed_only_due_to_grace = (marks < PASS_MARK) & (marks_after_grace >= PASS_MARK)
z_scores = (marks - marks.mean()) / marks.std()

print("T9a marks after +5 grace, capped at 100:", marks_after_grace)
print("T9b pass percentage after grace (%):", pass_percent)
print("T9c students who passed only because of grace (zero-based positions):",
      np.where(passed_only_due_to_grace)[0])
print("T9d z-scores:", np.round(z_scores, 3))
```

## Task 10 · An image is an array
- **(a)** Start from `np.zeros((8, 8), dtype=int)` and draw the "7" from the lecture with slicing (white = 255, grey = 128).
- **(b)** Flip it left-right.
- **(c)** Invert it (`255 - img`).
- **(d)** Crop rows 1–4 and columns 3–6.
- **(e)** Brighten by 80, capped at 255.
- **(f)** Print its shape, dtype and number of white pixels.

*Optional:* `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`

```python
# T10: Image is an array (a–f)
# Make an 8×8 black canvas; 0 = black, 255 = white.
img = np.zeros((8, 8), dtype=int)
img[1, 1:7] = 255
img[2:7, 5] = 255
img[4, 3:7] = 128

print("T10a original 7 image array:\n", img)
print("T10a shape:", img.shape, "| dtype:", img.dtype)

# b) Horizontal flip (reverse columns)
flipped = img[:, ::-1]
print("T10b horizontal flip:\n", flipped)

# c) Invert black and white/gray values
inverted = 255 - img
print("T10c inverted image:\n", inverted)

# d) Crop rows 1–4 and columns 3–6 (stop indexes excluded)
cropped = img[1:5, 3:7]
print("T10d cropped image:\n", cropped)

# e) Brighten by 80 but cap values at 255
brightened = np.clip(img + 80, 0, 255)
print("T10e brightened image:\n", brightened)

# f) Properties of original image
print("T10f original shape:", img.shape)
print("T10f original dtype:", img.dtype)
print("T10f white pixels (value exactly 255):", np.count_nonzero(img == 255))
```
