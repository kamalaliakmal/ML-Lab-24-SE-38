## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# T3: Clean the data in the required order

# (a) Missing values
print("T3a - Missing values before cleaning:")
print(df.isna().sum())

rows_before_missing = len(df)
# An ad without Brand or Model is not useful for identifying the car.
df = df.dropna(subset=["Brand", "Model"]).copy()
rows_removed_missing = rows_before_missing - len(df)

# Fill missing text categories with an explicit label. Do not replace
# missing Year or KMs Driven values; they remain NaN.
for column in ["Registered City", "Fuel", "Transaction Type"]:
    df[column] = df[column].fillna("Unknown")

print("T3a - Rows removed because Brand or Model was missing:", rows_removed_missing)
print("T3a - Missing values after this step:")
print(df.isna().sum())

# (b) Formatting: standardize city names and remove symbols from Price.
df["Registered City"] = df["Registered City"].astype("string").str.strip().str.title()
df["Price"] = (
    df["Price"].astype("string")
      .str.replace("Rs", "", regex=False)
      .str.replace(",", "", regex=False)
      .str.strip()
)
print("T3b - Example cleaned city names:")
print(df["Registered City"].head())
print("T3b - Example cleaned price values:")
print(df["Price"].head())

# (c) Convert text numbers into numeric data types.
# to_numeric with errors='raise' prevents silently losing unexpected values.
df["Price"] = pd.to_numeric(df["Price"], errors="raise").astype("int64")
df["Year"] = pd.to_numeric(df["Year"], errors="coerce").astype("Int64")
print("T3c - Data types after conversion:")
print(df[["Price", "Year"]].dtypes)

# (d) Correct inconsistent category spellings.
cities_before_correction = df["Registered City"].nunique()
df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})
cities_after_correction = df["Registered City"].nunique()
print("T3d - Different cities before spelling correction:", cities_before_correction)
print("T3d - Different cities after spelling correction:", cities_after_correction)

# (e) Remove exact duplicate records. Assign the returned DataFrame back.
exact_duplicates = df.duplicated().sum()
print("T3e - Exact duplicate ads:", exact_duplicates)
df = df.drop_duplicates().copy()
print("T3e - Row count after removing duplicates:", len(df))

# (f) Handle invalid mileage and fake prices using vectorized masks.
bad_km = (
    (df["KMs Driven"] < 100)
    | (df["KMs Driven"] > 1_000_000)
    | (df["KMs Driven"].isin([12345, 123456]))
)
df.loc[bad_km, "KMs Driven"] = np.nan

fake_price = df["Price"].isin([12_345_678, 87_654_321])
expensive_small_brands = (
    df["Brand"].isin(["Suzuki", "Daihatsu"]) & (df["Price"] > 5_000_000)
)
rows_before_invalid_price = len(df)
df = df.loc[~(fake_price | expensive_small_brands)].copy()
print("T3f - Rows removed for fake/invalid prices:", rows_before_invalid_price - len(df))
print("T3f - Invalid mileage values changed to NaN:", int(bad_km.sum()))

# (g) Required self-check from the lab handout.
print("T3g - Final shape:", df.shape)
print("T3g - Expected shape: (20484, 9)")
if df.shape != (20484, 9):
    print("T3g - WARNING: The shape does not match. Check that you are using the exact practice CSV supplied with this lab.")
```
